```python
import pandas as pd
import numpy as np
import json
```

```python
with open("../data/muslim.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
print(data["metadata"]["arabic"]["title"])
print(data["metadata"]["english"]["title"])
print(data["metadata"]["arabic"]["author"])
print(data["metadata"]["english"]["author"])
```

```python
arabic_title = data["metadata"]["arabic"]["title"]
english_title = data["metadata"]["english"]["title"]
arabic_author = data["metadata"]["arabic"]["author"]
english_author = data["metadata"]["english"]["author"]
```

```python
chapters = pd.DataFrame(data["chapters"])
chapters.drop("bookId", axis=1, inplace=True)
```

```python
chapters.head()
```

```python
hadiths = pd.json_normalize(data["hadiths"])
hadiths.head()
```

```python
hadiths.drop(["idInBook", "bookId"], axis=1, inplace=True)
hadiths.rename(columns={"id": "hadith_id", "chapterId": "chapter_id", "english.narrator": "narrator", "english.text": "english_text", "arabic": "arabic_text"}, inplace=True)
hadiths.head()
```

```python
print(chapters.columns)
print(hadiths.columns)
```

```python
df = pd.merge(hadiths, chapters, left_on="chapter_id", right_on="id", how="left")
df.drop("id", axis=1, inplace=True)
df.head()
```

```python
df.drop(["hadith_id", "chapter_id"], axis=1, inplace=True)
df.rename(columns={"arabic": "arabic_chapter_title", "english": "english_chapter_title"}, inplace=True)
df.head()
```

```python
df["arabic_title"] = arabic_title
df["english_title"] = english_title
df["arabic_author"] = arabic_author
df["english_author"] = english_author
```

```python
df.head()
```

```python
len(df)
```

```python
df.to_csv("../data/muslim.csv", index=False, encoding="utf-8-sig")
```
